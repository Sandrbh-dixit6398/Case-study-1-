```python
print("\n Welcome to the grocery store ")
a  = input("Enter the items: ").split()
print(a)
b = list(map(int,input("Enter quantity: ").split()))
print(b)
c = list(map(int,input("price per unit: ").split()))
print(c)
total = 0
for i in range(len(a)):
    total = total + b[i]*c[i]
print(f'Total cost of {a} is: {total}')
```
